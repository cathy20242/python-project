```python
#PRACTICAL 4
# WRITE A PROGRAM TO PERFORM THE FOLLOWING
#SET OPERATORS

#create /access/ update/ delete elements from set

# create set
numbers = {10 , 20 , 30 , 40 , 50 }
print ("set",numbers)
print("types",type(numbers)
)
```

```python
# create set  fro a string

letters = set("PYTHON")
print("characters:",letters)
print("types",type(letters))
```

```python
#access element from set

number={10 , 20 , 30 , 40 , 50}
for n in number:
    print(n)
```

```python
#membership testing

print ("is 30 present ?"in numbers)
print("is 100 present?"in numbers)
```

```python
#accessing elements from set using loop

print("set elements")
for n in numbers:
    print(n)
```

```python
#adding and updating set elements

numbers = {10 , 20 , 40}
print("original set:",numbers)
```

```python
#add one element

numbers = {10 , 20 , 40}
numbers.add(30)
print("updated set:",numbers)
```

```python
#add multiple element

numbers = {10 , 20 , 40}
numbers.update([30 , 50 , 60])
print("updated set:",numbers)
```

```python
#add multiples of 40

numbers = {10 , 20 , 40}
numbers.update([30 , 50 , 60] , {40 , 70})
print("updated set:",numbers)
```

```python
#add   4 even numbers between 10 and 40

numbers = {10 , 20 , 40}
numbers.update(range(40 , 46 , 2))
print("updated set:",numbers)
```

```python
# add odd numbers from a range of 10 to 30 in set numbers

numbers = {10 , 20 , 40}
numbers.update(range(11 , 30 , 2))
print("updated set:",numbers)
```

```python
#remove element from set

numbers = {10 , 20 , 40}
numbers.remove(20)
print("updated set:",numbers)
```

```python
#delete elements from set

numbers ={10 , 20 , 30 , 40 , 50}
print("original set:",numbers)
numbers.remove(30)
print("after remove (30):",numbers)
numbers.discard(40)
print("updated set:",numbers)
```
