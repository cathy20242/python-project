```python
#BINARY  SEARCH TREE

class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None

        root = node (8)


        root.left = node (3)
        root.right = node (10)
        root.left.left = node (1)
        root.left.right = node (6)
        root.right.right = node (14)
        root.right.right.left = node (13)

        print("root:"root.data)
        print("left child:"root.left.data)
        print("right child:"root.right.data)

        print("left child of 3:"root.left.left.data)
        print("left child of 3:"root.left.right.data)
        print("right of right:"root.right.right.data)
        print("right of right of right:"root.right.right.left.data)
```
