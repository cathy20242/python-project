```python
#dictionary of student marks

marks = {
    "python" : 85,
    "data structure": 90,
    "database": 78,
    "computer network": 90
}
print("student marks")
for subject, mark in marks.items():
    print(subject, ":", mark)
    print("average mark:", sum(marks.values())/len(marks))
```

```python
#set operations

set1 = { 10 , 20 , 30 , 40 , 50 }
set2 = {30 , 40 , 50 , 70}

print("set1 union set2")
print(set1.union(set2))

print("set1 intersection set2")
print(set1.intersection(set2))

print("set1 difference set2")
print(set1.difference(set2))
```

```python
# online shopping bill calculator

customer = input ("enter customer name: ")
amount  = input("enter purchase amount: ")
discount = 0
if amount > 10000:
    discount =amount*0.1
elif amount>= 20000:
  discount = amount *0.10
else :
  print("shop more to get discount")

final_bill = amount - discount
print("customer name:", customer)
print("purchase amount:", amount)
print("discount:", discount)
print("final bill:", final_bill)
```
